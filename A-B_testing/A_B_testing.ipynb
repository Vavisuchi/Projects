**CASE STUDY: OPTIMIZING LANDING PAGE VARIANTS FOR REVENUE**

**WHAT WE TESTED:**
three landing page designs (Vibrant, Cold, Heat) across 30,000 user sessions
for our Bluetooth speakers.

**KEY OBJECTIVES:**

*   Compare conversion and purchase rates across Vibrant, Cold, and Heat.
*   See how user engagement (time spent, pages visited) relates to checkouts.
*   Find out if specific segments (traffic sources, devices) prefer certain designs.
*   Calculate revenue per session to ensure we are maximizing profit, not just signups.


**STEPS:**

**Step 1-Statistical Testing:**

Run a Chi-Square test to check if conversion rate differences are statistically significant (p < 0.05) and use ANOVA to check differences in revenue and time spent.

**Step 2-Funnel & Behavior:**

Filter out bounces to analyze pages visited and time spent. See which variant keeps users engaged longer.

**Step 3-Segmentation:**

Cross-tabulate variants with traffic sources and device types to spot hidden preference trends.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats
import statsmodels.api as sm
```

```python
df = pd.read_csv("speakers_data.csv")
```

```python
df.head(10)
```

**statistical test to evaluate the A/B/C test performance of those landing page variants (Vibrant, Cold, Heat).**

```python
print("Dataset Shape:", df.shape)

# 1. Basic Inspection: Check missing values and data types
print("\n--- Missing Values Check ---")
print(df.isnull().sum()[df.isnull().sum() > 0])

# 2. A/B/C Test Performance: Conversion Rate by Variant Group
print("\n--- Conversion Rate & Revenue by Landing Page Variant ---")
variant_summary = df.groupby('variant_group').agg(
    total_sessions=('session_id', 'count'),
    total_conversions=('conversion_flag', 'sum'),
    conversion_rate=('conversion_flag', 'mean'),
    avg_revenue=('revenue_$', 'mean'),
    total_revenue=('revenue_$', 'sum')
).reset_index()

# Convert conversion rate to percentage for readability
variant_summary['conversion_rate (%)'] = (variant_summary['conversion_rate'] * 100).round(2)
display(variant_summary)

# 3. Statistical Significance Test (Chi-Square Test for Independence)
print("\n--- Chi-Square Test: Variant Group vs. Conversion Flag ---")
contingency_table = pd.crosstab(df['variant_group'], df['conversion_flag'])
chi2, p_val, dof, expected = stats.chi2_contingency(contingency_table)

print(f"Chi2 Statistic: {chi2:.4f}")
print(f"p-value: {p_val:.4f}")

if p_val < 0.05:
    print("Result: Statistically significant difference in conversion rates across variants! (Reject Null Hypothesis)")
else:
    print("Result: No statistically significant difference in conversion rates across variants. (Fail to Reject Null Hypothesis)")
```

```python
# 1. Behavioral Engagement by Variant Group
print("--- Behavioral Engagement by Variant ---")
engagement_summary = df.groupby('variant_group').agg(
    avg_time_spent=('time_spent', 'mean'),
    avg_pages_visited=('pages_visited', 'mean'),
    bounce_rate=('bounce_flag', 'mean')
).reset_index()

engagement_summary['bounce_rate (%)'] = (engagement_summary['bounce_rate'] * 100).round(2)
display(engagement_summary)

# 2. Conversion Rate by Traffic Source & Variant Group
print("\n--- Conversion Rate by Traffic Source & Variant ---")
traffic_variant = df.groupby(['variant_group', 'traffic_source']).agg(
    sessions=('session_id', 'count'),
    conversion_rate=('conversion_flag', 'mean'),
    avg_revenue=('revenue_$', 'mean')
).reset_index()

traffic_variant['conversion_rate (%)'] = (traffic_variant['conversion_rate'] * 100).round(2)
display(traffic_variant)
```

```python
#Funnel Analysis

# Funnel Analysis by Variant Group
print("--- Funnel Analysis by Landing Page Variant ---")

# We create step flags for clarity
df['engaged_flag'] = 1 - df['bounce_flag'] # 1 if they didn't bounce
df['signed_in_flag'] = np.where(df['sign_in'] == 'Email', 1, 0)

funnel_summary = df.groupby('variant_group').agg(
    total_sessions=('session_id', 'count'),
    engaged_sessions=('engaged_flag', 'sum'),
    signed_in_users=('signed_in_flag', 'sum'),
    converted_users=('conversion_flag', 'sum')
).reset_index()

# Calculate drop-off/conversion rates from total sessions
funnel_summary['engagement_rate (%)'] = (funnel_summary['engaged_sessions'] / funnel_summary['total_sessions'] * 100).round(2)
funnel_summary['signin_rate (%)'] = (funnel_summary['signed_in_users'] / funnel_summary['total_sessions'] * 100).round(2)
funnel_summary['overall_conversion_rate (%)'] = (funnel_summary['converted_users'] / funnel_summary['total_sessions'] * 100).round(2)

display(funnel_summary[['variant_group', 'total_sessions', 'engaged_sessions', 'signed_in_users', 'converted_users', 'overall_conversion_rate (%)']])
```
