So instead, let's directly answer the business question:

Does price elasticity differ across metro areas?

To answer that, we need an interaction effect between log_Price and geography.

```python
# Keep only the 10 individual metro areas
metro_names = [
    'Atlanta-Sandy Springs-Roswell  GA',
    'Boston-Cambridge-Newton  MA-NH',
    'Chicago-Naperville-Elgin  IL-IN-WI',
    'Dallas-Fort Worth-Arlington  TX',
    'Detroit-Warren-Dearborn  MI',
    'Houston-The Woodlands-Sugar Land  TX',
    'Los Angeles-Long Beach-Anaheim  CA',
    'Miami-Fort Lauderdale-West Palm Beach  FL',
    'New York-Newark-Jersey City  NY-NJ-PA',
    'Philadelphia-Camden-Wilmington  PA-NJ-DE-MD'
]

metro_df = df[df["Metroregion_name"].isin(metro_names)].copy()

print("Observations:", len(metro_df))
print("Metro areas:", metro_df["Metroregion_name"].nunique())
print(metro_df["Metroregion_name"].value_counts())
```

For each metro, we'll estimate:

log(Q)=α+βlog(P)+ϵ

where β is the price elasticity for that metro.

```python
# Metro-level price elasticity

metro_results = []

# Keep only the 10 actual metropolitan areas
metro_df = df[
    ~df["Metroregion_name"].isin([
        "National",
        "Census Region 1: Northeast",
        "Census Region 2: Midwest",
        "Census Region 3: South",
        "Census Region 4: West"
    ])
].copy()

for metro in metro_df["Metroregion_name"].unique():

    metro_data = metro_df[
        metro_df["Metroregion_name"] == metro
    ]

    X = sm.add_constant(metro_data["log_Price"])
    y = metro_data["log_Quantity"]

    model = sm.OLS(y, X).fit()

    metro_results.append({
        "Metro": metro,
        "Elasticity": model.params["log_Price"],
        "P_value": model.pvalues["log_Price"],
        "R_squared": model.rsquared
    })

metro_elasticity = pd.DataFrame(metro_results)

metro_elasticity = metro_elasticity.sort_values("Elasticity")

print(metro_elasticity)
```

The R² values are roughly 0.087–0.127.

For example:

*This means price alone explains roughly 9–13% of the variation in log quantity within these metro-level regressions.*

*Compare price and quantity within each metro while accounting for the fact that different food categories have very different typical quantities, and that purchasing can change over time.*

```python
# Metro-level elasticity controlling for food category and time

metro_controlled_results = []

for metro in metro_df["Metroregion_name"].unique():

    metro_data = metro_df[
        metro_df["Metroregion_name"] == metro
    ].copy()

    # Food category controls
    food_dummies = pd.get_dummies(
        metro_data["EFPG_name"],
        prefix="Food",
        drop_first=True
    ).astype(int)

    # Regression variables
    X_metro = pd.concat(
        [
            metro_data[["log_Price", "Year", "Month"]],
            food_dummies
        ],
        axis=1
    )

    X_metro = sm.add_constant(X_metro)

    y_metro = metro_data["log_Quantity"]

    model = sm.OLS(y_metro, X_metro).fit()

    metro_controlled_results.append({
        "Metro": metro,
        "Elasticity": model.params["log_Price"],
        "P_value": model.pvalues["log_Price"],
        "R_squared": model.rsquared
    })

metro_controlled_elasticity = pd.DataFrame(
    metro_controlled_results
)

metro_controlled_elasticity = (
    metro_controlled_elasticity
    .sort_values("Elasticity")
)

print(metro_controlled_elasticity)
```

So we can absolutely say:

"The estimated elasticities vary across metros."

That's a descriptive comparison.

What we don't know yet

Suppose New York is −1.805 and Los Angeles is −1.007.

The difference is:

−1.805−(−1.007)=−0.798

That's a fairly large numerical difference.

But statistical analysis asks a different question:

*Could this difference simply be due to sampling/noise, or is there evidence that the underlying price elasticity really differs between these metros?*

That's what the interaction model tests.

```python
# Interaction model:
# Does price elasticity differ across metropolitan areas?

# Make a copy of the 10-metro dataset
interaction_df = metro_df.copy()

# Create metro dummy variables
metro_dummies = pd.get_dummies(
    interaction_df["Metroregion_name"],
    prefix="Metro",
    drop_first=True
).astype(int)

# Create price × metro interaction terms
interaction_terms = metro_dummies.multiply(
    interaction_df["log_Price"],
    axis=0
)

interaction_terms.columns = [
    col + "_x_logPrice"
    for col in interaction_terms.columns
]

# Create food-category dummy variables
food_dummies = pd.get_dummies(
    interaction_df["EFPG_name"],
    prefix="Food",
    drop_first=True
).astype(int)

# Build the full model
X_interaction = pd.concat(
    [
        interaction_df[["log_Price", "Year", "Month"]],
        food_dummies,
        metro_dummies,
        interaction_terms
    ],
    axis=1
)

X_interaction = sm.add_constant(X_interaction)

y_interaction = interaction_df["log_Quantity"]

# Run regression
interaction_model = sm.OLS(
    y_interaction,
    X_interaction
).fit()

print(interaction_model.summary())
```

```python
# Identify all metro × price interaction variables
interaction_vars = [
    col for col in X_interaction.columns
    if "_x_logPrice" in col
]

print("Interaction variables:")
print(interaction_vars)

print("\nNumber of interaction variables:", len(interaction_vars))
```

```python
# Joint F-test of all metro × price interaction effects

hypotheses = " = 0, ".join(interaction_vars) + " = 0"

joint_test = interaction_model.f_test(hypotheses)

print(joint_test)
```

We tested:

H₀: All 9 metro × price interaction effects are zero.

In ELI5 terms:

"Suppose Atlanta's price elasticity applies equally to every other metro. Is that assumption reasonable?"

Therefore, we reject the null hypothesis.

So we have very strong statistical evidence that:

***Price elasticity is not the same across all metro areas.***

In other words, geography matters for price sensitivity.

```python
# Calculate final price elasticity for each metro

baseline_elasticity = interaction_model.params["log_Price"]

metro_elasticities = []

# Reference metro: Atlanta
metro_elasticities.append({
    "Metro": "Atlanta-Sandy Springs-Roswell  GA",
    "Elasticity": baseline_elasticity
})

# Other metros
for interaction_var in interaction_vars:

    metro_name = interaction_var.replace("_x_logPrice", "")

    interaction_effect = interaction_model.params[interaction_var]

    final_elasticity = baseline_elasticity + interaction_effect

    metro_elasticities.append({
        "Metro": metro_name.replace("Metro_", ""),
        "Elasticity": final_elasticity
    })

metro_elasticity_results = pd.DataFrame(metro_elasticities)

# Sort from most negative to least negative
metro_elasticity_results = metro_elasticity_results.sort_values(
    "Elasticity"
)

print(metro_elasticity_results)
```

One important issue

Since we're doing 45 comparisons, we shouldn't simply look at each raw p-value at 0.05. That creates a multiple-comparisons problem.

We'll therefore use a correction such as Holm's correction when interpreting the pairwise tests.

```python
from itertools import combinations
from statsmodels.stats.multitest import multipletests
import pandas as pd

# Get the 9 interaction terms
interaction_vars = [
    col for col in interaction_model.params.index
    if "_x_logPrice" in col
]

# Create all unique pairs of metros
pairs = list(combinations(interaction_vars, 2))

results = []

for var1, var2 in pairs:

    # Test whether the two metro interaction coefficients are equal
    test = interaction_model.t_test(f"{var1} = {var2}")

    results.append({
        "Metro_A": var1.replace("Metro_", "").replace("_x_logPrice", ""),
        "Metro_B": var2.replace("Metro_", "").replace("_x_logPrice", ""),
        "Difference": float(test.effect),
        "P_value": float(test.pvalue)
    })

pairwise_results = pd.DataFrame(results)

# Correct the 45 p-values for multiple comparisons
pairwise_results["P_value_adjusted"] = multipletests(
    pairwise_results["P_value"],
    method="holm"
)[1]

# Statistical significance after correction
pairwise_results["Significant"] = (
    pairwise_results["P_value_adjusted"] < 0.05
)

# Show the strongest evidence first
pairwise_results = pairwise_results.sort_values(
    "P_value_adjusted"
)

print(pairwise_results.to_string(index=False))
```

SUMMARIZING THIS PAIRWISE COMPARISION:
This pairwise test now tells us which specific metros differ from each other.

There are 45 possible metro pairs among 10 metros.
After the multiple-comparison adjustment, 27 pairs are statistically significantly different.
18 pairs are not significantly different.

So there is substantial geographic variation, but not every metro differs from every other metro.

Overall interpretation

So our geographic analysis now has three levels of evidence:

Separate metro regressions → gave us the elasticity estimates for each metro.
Interaction model + F-test → formally established that elasticity varies across metros overall.
Pairwise comparisons → identified which metro pairs are significantly different.

And that's much stronger than simply saying "New York has −1.37 and LA has −1.29, therefore they're different." We have now statistically tested that difference.

# Pricing Strategy to Achieve a 5% Revenue Increase Across 10 U.S. Metropolitan Areas

## Executive Summary

This analysis examines how food prices are associated with purchasing quantities across **10 major U.S. metropolitan areas**. Using a log-log demand model with controls for food category and time, we estimated metro-specific price elasticities and formally tested whether price sensitivity differs geographically.

The results show that **all 10 metropolitan areas have negative price elasticities**, meaning that higher prices are associated with lower purchasing quantities. Estimated elasticities range from approximately **−1.28 to −1.37**.

A joint statistical test confirms that price elasticity differs significantly across metropolitan areas (**F = 33.70, p < 0.001**). Therefore, a single national pricing strategy is unlikely to be optimal.

For a retailer targeting a **5% increase in total revenue across the 10 metropolitan areas**, the recommended strategy is **not to simply raise prices by 5%**. Because demand is elastic, a price increase would reduce quantity enough to potentially reduce revenue.

Instead, the retailer should use **targeted price reductions on selected products/categories where the resulting increase in quantity is expected to outweigh the lower price**, while using differentiated pricing across metropolitan markets.

---

# 1. Research Question

**How can a U.S. food retailer use geographically differentiated price strategies to increase total revenue by 5% across 10 major metropolitan markets while accounting for differences in consumer price sensitivity?**

---

# 2. Key Findings

### 2.1 Price has a negative relationship with purchasing quantity

The estimated metro-specific elasticities are:

| Metropolitan Area | Estimated Price Elasticity |
| ----------------- | -------------------------: |
| New York          |                 **−1.372** |
| Dallas            |                 **−1.366** |
| Houston           |                 **−1.366** |

| Miami             |                 **−1.325** |

All elasticities are below −1, indicating **elastic demand** in the estimated model.

Conversely, a 1% price reduction is associated with approximately a **1.37% increase in quantity**, based on the estimated relationship.

# 3. Why a Uniform Pricing Strategy Is Not Recommended

The interaction model shows that elasticity differs significantly across metropolitan areas.

**F = 33.70**

**p = 7.08 × 10⁻⁶⁰**

Therefore, we reject the hypothesis that all metropolitan areas have the same price elasticity.

The pairwise analysis further shows that **27 of the 45 metro pairs have statistically significant differences** after Holm multiple-comparison correction.

This supports a **differentiated pricing strategy rather than one national price change**.

# 4. Revenue Implication

Revenue can be represented as:

[
R=P\times Q
]

If price changes by a percentage (\Delta P), the approximate percentage change in quantity is:

[
%\Delta Q \approx \beta \times %\Delta P
]

Therefore:

[
%\Delta R \approx %\Delta P+%\Delta Q
]

or:

[
%\Delta R \approx (1+\beta)%\Delta P
]

Because the estimated elasticities are approximately −1.28 to −1.37:

[
1+\beta < 0
]

This means that **a price increase is expected to reduce revenue**, while a price reduction can increase revenue within the range where the elasticity approximation is appropriate.

For example, in New York:

[
\beta=-1.372
]

A 1% price reduction would be associated with approximately:

[
1.372%
]

increase in quantity.

The approximate revenue effect would therefore be:

[
-1%+1.372%=+0.372%
]

Thus, a modest price reduction could increase revenue because the increase in quantity is larger than the decrease in price.

# 5. Recommended Strategy for a 5% Revenue Increase

## Recommendation: Use targeted price reductions rather than broad price increases

The retailer should **not reduce every product by the same percentage**.

Instead, the recommended strategy is:

### 1. Identify high-volume, price-sensitive food categories

Focus discounts on categories where:

* purchasing volume is high;
* price sensitivity is high;
* products are frequently purchased;
* consumers can easily increase quantity when prices fall.

These categories have the greatest potential to generate additional volume.

### 2. Apply moderate price reductions

Rather than immediately implementing very large discounts, begin with **small-to-moderate price reductions**, such as **2–5%**, and monitor quantity and revenue responses.

For example, using an elasticity around −1.30:

| Price change | Approx. quantity change | Approx. revenue change* |
| -----------: | ----------------------: | ----------------------: |
|          −2% |                   +2.6% |               **+0.6%** |
|          −3% |                   +3.9% |               **+0.9%** |
|          −4% |                   +5.2% |               **+1.2%** |
|          −5% |                   +6.5% |               **+1.5%** |

*Approximation based on the linear elasticity relationship.

This demonstrates why **small price reductions can potentially increase revenue when demand is elastic**.

# 8. How the Retailer Can Reach the 5% Target

The 5% target should be treated as a **portfolio-level revenue objective**, rather than requiring every metro and every food category to achieve 5%.

The retailer should:

**Step 1:** Rank food categories by sales volume.

**Step 2:** Estimate elasticity separately by food category and metro.

**Step 3:** Identify products where demand is sufficiently elastic to make price reductions revenue-enhancing.

**Step 4:** Apply small price reductions, initially around **2–5%**, to selected high-volume products.

**Step 5:** Measure the resulting quantity and revenue changes.

**Step 6:** Shift more promotional activity toward markets/categories generating the strongest revenue response.

**Step 7:** Continue adjusting prices until the combined revenue increase approaches the **5% target**.

# 9. Important Limitation

The current analysis provides strong evidence of **association between price and quantity**, but it does not by itself establish a fully causal demand curve.

Therefore, the retailer should **not interpret the elasticity estimates as a guarantee** that a 5% price reduction will produce exactly a 6.5% increase in quantity.

For example, the estimate of −1.30 suggests:

> 5% lower price → approximately 6.5% higher quantity

but the actual response could differ because of promotions, competitors, seasonality, consumer income, product substitution, stock availability, and other factors.

The best implementation would therefore use **controlled price experiments/A-B tests** before applying the strategy across the entire portfolio.

# 10. Final Recommendation

### Recommended pricing strategy

**The retailer should pursue a geographically differentiated, targeted price-reduction strategy rather than a uniform price increase.**

Because estimated price elasticities across the 10 metropolitan areas are approximately **−1.28 to −1.37**, demand is elastic in the estimated models. Consequently, carefully targeted price reductions can potentially increase quantity by more than the proportional reduction in price, generating additional revenue.

The retailer should initially test **2–5% price reductions on high-volume, highly price-sensitive food categories**, with larger initial reductions in relatively more price-sensitive markets such as **New York, Dallas, and Houston**, and smaller reductions in markets such as **Philadelphia, Los Angeles, and Chicago**.

The ultimate objective should be to construct a portfolio of price changes that produces a **5% increase in total revenue across the 10 metropolitan markets**, rather than applying a blanket 5% price change.

**In short:**

> **Do not raise prices to increase revenue. Use targeted, moderate price reductions where demand is elastic, differentiate the strategy by metro and food category, and continuously test the revenue response until the portfolio reaches the 5% revenue-growth target.**
