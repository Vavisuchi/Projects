```python
#changing the date format
df['start_date']=pd.to_datetime(df['start_date'])
df['end_date']=pd.to_datetime(df['end_date'])
```

**Which claim types cost the most?**

For every type of medical claim, calculate the average hospital spending per episode, then rank them from highest spending to lowest.

```python
df.groupby(
    "claim_type"
)["avg_spndg_per_ep_hospital"].mean().sort_values(
    ascending=False
)
```

**Which hospitals spend the most?**

```python
df.groupby(["facility_id","facility_name"])["avg_spndg_per_ep_hospital"].mean().sort_values(ascending=False).head(10)
```

**Where is this hospital above national average?**

```python
df["difference_vs_national"] = (
    df["avg_spndg_per_ep_hospital"]
    -
    df["avg_spndg_per_ep_national"]
)
```

```python
missing = df.isnull().sum()

missing[missing > 0]
```

```python
percentage_columns = [
    "percent_of_spndg_hospital",
    "percent_of_spndg_state",
    "percent_of_spndg_national"
]

for col in percentage_columns:
    # Check if the column is of object type (string-like) before applying .str accessor
    if df[col].dtype == 'object':
        df[col] = (
            df[col]
            .str.replace("%", "", regex=False)
            .astype(float)
        )
    # If it's already a numeric type but not float, ensure it's float
    elif not pd.api.types.is_float_dtype(df[col]):
        df[col] = df[col].astype(float)
```

```python
df[percentage_columns].head()
```

```python
#creating copy of dataset

df_clean=df.copy()
```

```python
df_clean[percentage_columns].head()
```

**KPI 1: How many hospitals are we benchmarking?**

```python
total_hospitals=df_clean["facility_id"].nunique()
print(total_hospitals)
```

**KPI 2 – States Covered**

```python
total_states=df_clean["state"].nunique()
print(total_states)
```

***KPI 3 – Average Episode Cost (National Dataset)***

```python
episode_totals=df_clean[(df_clean["period"]=="Complete Episode")& (df_clean["claim_type"]=="Total")]
```

```python
episode_totals.head()
```

**KPI 4 – Average Episode Cost Across All Hospitals**

```python
average_episode_cost = episode_totals["avg_spndg_per_ep_hospital"].mean()
print(average_episode_cost)
```

**Highest-Cost Hospital**

```python
highest_cost_hospital = episode_totals.loc[
    episode_totals["avg_spndg_per_ep_hospital"].idxmax()
]

highest_cost_hospital
```

**State-Level Benchmarking**

```python
state_summary = (
    episode_totals
    .groupby("state")["avg_spndg_per_ep_hospital"]
    .mean()
    .reset_index()
)
```

```python
state_summary.head()
```

```python
state_summary = state_summary.sort_values(
    by="avg_spndg_per_ep_hospital",
    ascending=False
)
```
