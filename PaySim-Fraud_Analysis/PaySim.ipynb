

> **How can the mobile money provider reduce fraud-related financial losses and improve profitability by identifying the customers and transaction types with the highest fraud risk?**


```python
# ============================================================
# A. BUSINESS UNDERSTANDING
# ============================================================

# Client:
# Head of Risk / Fraud Management at a mobile money provider
#
# Business problem:
# Fraud causes financial losses and can reduce profitability.
#
# Business question:
# How can the mobile money provider reduce fraud-related
# financial losses and improve profitability by identifying
# the customers and transaction types with the highest
# fraud risk?


# ============================================================
# B. DATA UNDERSTANDING
# ============================================================

# ------------------------------------------------------------
# B1. Import libraries
# ------------------------------------------------------------

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Optional: improve display
pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", "{:,.2f}".format)


# ------------------------------------------------------------
# B2. Load the dataset
# ------------------------------------------------------------

df = pd.read_csv("PaySim_Data.csv")


# ------------------------------------------------------------
# B3. Preview the dataset
# ------------------------------------------------------------

print("First 10 rows:")
display(df.head(10))


# ------------------------------------------------------------
# B4. Check the size of the dataset
# ------------------------------------------------------------

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])


# ------------------------------------------------------------
# B5. Check column names and data types
# ------------------------------------------------------------

print("\nDataset information:")
df.info()


# ------------------------------------------------------------
# B6. Check for missing values
# ------------------------------------------------------------

print("\nMissing values:")
print(df.isnull().sum())


# ------------------------------------------------------------
# B7. Check duplicate records
# ------------------------------------------------------------

duplicate_count = df.duplicated().sum()

print("\nNumber of duplicate records:", duplicate_count)


# ============================================================
# B8. Examine the target variable: isFraud
# ============================================================

# Count fraudulent vs legitimate transactions

fraud_counts = df["isFraud"].value_counts()

print("\nFraudulent vs legitimate transactions:")
print(fraud_counts)


# ------------------------------------------------------------
# B9. Fraud percentage
# ------------------------------------------------------------

fraud_percentage = (
    df["isFraud"]
    .value_counts(normalize=True)
    * 100
)

print("\nFraud percentage:")
print(fraud_percentage)


# ------------------------------------------------------------
# B10. Number of fraudulent and legitimate transactions
# ------------------------------------------------------------

total_transactions = len(df)

fraud_transactions = df["isFraud"].sum()

legitimate_transactions = (
    total_transactions - fraud_transactions
)

print("\nTotal transactions:", total_transactions)
print("Fraudulent transactions:", fraud_transactions)
print("Legitimate transactions:", legitimate_transactions)

print(
    "Fraud rate:",
    round(
        fraud_transactions / total_transactions * 100,
        4
    ),
    "%"
)


# ============================================================
# C. EXPLORATORY DATA ANALYSIS
# ============================================================


# ------------------------------------------------------------
# C1. Total transaction value
# ------------------------------------------------------------

total_transaction_value = df["amount"].sum()

print(
    "\nTotal transaction value:",
    f"${total_transaction_value:,.2f}"
)


# ------------------------------------------------------------
# C2. Average transaction amount
# ------------------------------------------------------------

average_transaction_amount = df["amount"].mean()

print(
    "Average transaction amount:",
    f"${average_transaction_amount:,.2f}"
)


# ============================================================
# C3. Total fraudulent transaction value
# ============================================================

fraudulent_transaction_value = df.loc[
    df["isFraud"] == 1,
    "amount"
].sum()

print(
    "\nTotal fraudulent transaction value:",
    f"${fraudulent_transaction_value:,.2f}"
)


# ------------------------------------------------------------
# C4. Total legitimate transaction value
# ------------------------------------------------------------

legitimate_transaction_value = df.loc[
    df["isFraud"] == 0,
    "amount"
].sum()

print(
    "Total legitimate transaction value:",
    f"${legitimate_transaction_value:,.2f}"
)


# ------------------------------------------------------------
# C5. Fraud value as percentage of total transaction value
# ------------------------------------------------------------

fraud_value_percentage = (
    fraudulent_transaction_value
    / total_transaction_value
) * 100

print(
    "Fraud value as % of total transaction value:",
    round(fraud_value_percentage, 4),
    "%"
)


# ============================================================
# C6. Fraud by transaction type
# ============================================================

# Total transactions by transaction type

total_by_type = (
    df.groupby("type")
    .size()
)

print("\nTotal transactions by type:")
print(total_by_type)


# ------------------------------------------------------------
# C7. Fraudulent transactions by transaction type
# ------------------------------------------------------------

fraud_by_type = (
    df[df["isFraud"] == 1]
    .groupby("type")
    .size()
    .sort_values(ascending=False)
)

print("\nFraudulent transactions by type:")
print(fraud_by_type)


# ============================================================
# C8. Fraud rate by transaction type
# ============================================================

fraud_by_type = (
    df.groupby("type")["isFraud"]
    .sum()
)

fraud_rate_by_type = (
    fraud_by_type
    / total_by_type
    * 100
).sort_values(ascending=False)


# Create summary table

fraud_type_analysis = pd.DataFrame({
    "total_transactions": total_by_type,
    "fraudulent_transactions": fraud_by_type,
    "fraud_rate_percent": fraud_rate_by_type
})

fraud_type_analysis = (
    fraud_type_analysis
    .sort_values(
        "fraud_rate_percent",
        ascending=False
    )
)

print("\nFraud analysis by transaction type:")
display(fraud_type_analysis)


# ============================================================
# C9. Visualize fraud rate by transaction type
# ============================================================

plt.figure(figsize=(10, 6))

sns.barplot(
    x=fraud_type_analysis.index,
    y=fraud_type_analysis["fraud_rate_percent"]
)

plt.title("Fraud Rate by Transaction Type")
plt.xlabel("Transaction Type")
plt.ylabel("Fraud Rate (%)")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()


# ============================================================
# C10. Total fraudulent amount by transaction type
# ============================================================

fraud_amount_by_type = (
    df[df["isFraud"] == 1]
    .groupby("type")["amount"]
    .sum()
    .sort_values(ascending=False)
)

print("\nTotal fraudulent amount by transaction type:")
print(fraud_amount_by_type)


# ------------------------------------------------------------
# C11. Visualize fraudulent amount by transaction type
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

sns.barplot(
    x=fraud_amount_by_type.index,
    y=fraud_amount_by_type.values
)

plt.title(
    "Total Fraudulent Transaction Value by Type"
)

plt.xlabel("Transaction Type")
plt.ylabel("Fraudulent Amount")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()


# ============================================================
# C12. Average fraudulent transaction amount
# ============================================================

average_fraud_amount = df.loc[
    df["isFraud"] == 1,
    "amount"
].mean()

print(
    "\nAverage fraudulent transaction amount:",
    f"${average_fraud_amount:,.2f}"
)


# ------------------------------------------------------------
# C13. Average legitimate transaction amount
# ------------------------------------------------------------

average_legitimate_amount = df.loc[
    df["isFraud"] == 0,
    "amount"
].mean()

print(
    "Average legitimate transaction amount:",
    f"${average_legitimate_amount:,.2f}"
)


# ------------------------------------------------------------
# C14. Difference between average fraud and legitimate amount
# ------------------------------------------------------------

amount_difference = (
    average_fraud_amount
    - average_legitimate_amount
)

print(
    "Difference:",
    f"${amount_difference:,.2f}"
)


# ============================================================
# C15. Transaction amount: fraud vs legitimate
# ============================================================

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="isFraud",
    y="amount"
)

plt.title(
    "Transaction Amount: Fraud vs Legitimate"
)

plt.xlabel(
    "Fraud Status (0 = Legitimate, 1 = Fraud)"
)

plt.ylabel("Transaction Amount")

# Limit extreme values so the chart is readable
plt.ylim(
    0,
    df["amount"].quantile(0.99)
)

plt.tight_layout()
plt.show()


# ============================================================
# C16. Fraudulent transactions over time
# ============================================================

fraud_by_step = (
    df[df["isFraud"] == 1]
    .groupby("step")
    .size()
)

print("\nFraudulent transactions by step:")
print(fraud_by_step.head())


# ------------------------------------------------------------
# C17. Plot fraudulent transactions over time
# ------------------------------------------------------------

plt.figure(figsize=(12, 6))

plt.plot(
    fraud_by_step.index,
    fraud_by_step.values
)

plt.title(
    "Fraudulent Transactions Over Time"
)

plt.xlabel("Step")
plt.ylabel("Number of Fraudulent Transactions")

plt.tight_layout()
plt.show()


# ============================================================
# C18. Fraud rate over time
# ============================================================

# Total transactions by step

total_by_step = (
    df.groupby("step")
    .size()
)


# Fraudulent transactions by step

fraud_by_step = (
    df.groupby("step")["isFraud"]
    .sum()
)


# Fraud rate by step

fraud_rate_by_step = (
    fraud_by_step
    / total_by_step
    * 100
)

print("\nFraud rate by step:")
print(fraud_rate_by_step.head())


# ------------------------------------------------------------
# C19. Plot fraud rate over time
# ------------------------------------------------------------

plt.figure(figsize=(12, 6))

plt.plot(
    fraud_rate_by_step.index,
    fraud_rate_by_step.values
)

plt.title("Fraud Rate Over Time")

plt.xlabel("Step")
plt.ylabel("Fraud Rate (%)")

plt.tight_layout()
plt.show()


# ============================================================
# C20. Customers/accounts involved in fraudulent transactions
# ============================================================

# Fraudulent originator accounts

fraud_originators = (
    df[df["isFraud"] == 1]
    .groupby("nameOrig")
    .agg(
        fraud_transactions=("isFraud", "sum"),
        total_fraud_amount=("amount", "sum")
    )
    .sort_values(
        "total_fraud_amount",
        ascending=False
    )
)

print(
    "\nTop accounts associated with fraudulent transactions "
    "as originators:"
)

display(fraud_originators.head(10))


# ============================================================
# C21. Fraudulent destination accounts
# ============================================================

fraud_destinations = (
    df[df["isFraud"] == 1]
    .groupby("nameDest")
    .agg(
        fraud_transactions=("isFraud", "sum"),
        total_fraud_amount=("amount", "sum")
    )
    .sort_values(
        "total_fraud_amount",
        ascending=False
    )
)

print(
    "\nTop accounts associated with fraudulent transactions "
    "as destinations:"
)

display(fraud_destinations.head(10))


# ============================================================
# C22. isFlaggedFraud vs actual fraud
# ============================================================

flagged_vs_actual = pd.crosstab(
    df["isFlaggedFraud"],
    df["isFraud"],
    rownames=["System Flagged"],
    colnames=["Actual Fraud"]
)

print("\nisFlaggedFraud vs Actual Fraud:")
display(flagged_vs_actual)


# ============================================================
# C23. Fraud detection rate of the existing flag
# ============================================================

# Actual fraudulent transactions that were flagged

fraud_flagged = df[
    (df["isFraud"] == 1) &
    (df["isFlaggedFraud"] == 1)
].shape[0]


# Total fraudulent transactions

total_fraud = df["isFraud"].sum()


# Percentage of fraud detected by the flag

fraud_detection_rate = (
    fraud_flagged
    / total_fraud
    * 100
)

print(
    "\nFraudulent transactions flagged:",
    fraud_flagged
)

print(
    "Total fraudulent transactions:",
    total_fraud
)

print(
    "Fraud detection rate:",
    round(fraud_detection_rate, 2),
    "%"
)


# ============================================================
# C24. Potential financial savings
# ============================================================

# Test different fraud prevention scenarios

prevention_rates = [
    0.10,
    0.25,
    0.50,
    0.75,
    1.00
]

print("\nPotential savings from preventing fraud:")

for rate in prevention_rates:

    savings = (
        fraudulent_transaction_value
        * rate
    )

    print(
        f"{rate * 100:.0f}% fraud prevention "
        f"-> Potential savings: "
        f"${savings:,.2f}"
    )


# ============================================================
# C25. Create a potential savings table
# ============================================================

savings_data = []

for rate in prevention_rates:

    savings = (
        fraudulent_transaction_value
        * rate
    )

    savings_data.append({
        "fraud_prevention_rate": f"{rate * 100:.0f}%",
        "potential_savings": savings
    })

savings_df = pd.DataFrame(savings_data)

print("\nPotential savings table:")
display(savings_df)


# ============================================================
# C26. Final EDA summary numbers
# ============================================================

print("\n" + "=" * 60)
print("FINAL EDA SUMMARY")
print("=" * 60)

print(
    "Total transactions:",
    f"{total_transactions:,}"
)

print(
    "Fraudulent transactions:",
    f"{fraud_transactions:,}"
)

print(
    "Fraud rate:",
    f"{fraud_transactions / total_transactions * 100:.4f}%"
)

print(
    "Total transaction value:",
    f"${total_transaction_value:,.2f}"
)

print(
    "Total fraudulent transaction value:",
    f"${fraudulent_transaction_value:,.2f}"
)

print(
    "Fraud value % of total transaction value:",
    f"{fraud_value_percentage:.4f}%"
)

print(
    "Average fraudulent transaction:",
    f"${average_fraud_amount:,.2f}"
)

print(
    "Average legitimate transaction:",
    f"${average_legitimate_amount:,.2f}"
)

print(
    "Fraud detection rate of isFlaggedFraud:",
    f"{fraud_detection_rate:.2f}%"
)

print("=" * 60)
```

Fraud is rare, but that doesn't necessarily mean the financial impact is small.

The number of fraudulent transactions alone cannot tell us how much money the company is losing.

For example, 8,213 fraudulent transactions could represent relatively small amounts—or they could represent millions/billions in financial losses.



Insight 1: Fraud is low in volume but high in financial impact
Only 0.1291% of all transactions are fraudulent.
However, fraudulent transactions represent 1.0535% of total transaction value.
This means fraud is disproportionately affecting the company's financial value compared with its transaction volume.

Business implication: Fraud prevention should focus on financial impact, not just the number of fraudulent transactions.

Insight 2: Fraudulent transactions are significantly higher in value
Average fraudulent transaction: $1,467,967
Average legitimate transaction: $178,197
A fraudulent transaction is approximately 8.24 times larger than an average legitimate transaction.

Business implication: High-value transactions should receive stronger fraud monitoring and additional verification.

Insight 3: The potential financial exposure is substantial
Total transaction value: approximately $1.144 trillion
Fraudulent transaction value: approximately $12.06 billion

Business implication: Even though fraud is relatively rare, reducing fraudulent transaction value could produce substantial financial benefits and protect profitability.

Insight 4: Existing fraud flagging has very limited coverage
The isFlaggedFraud indicator detected only 0.19% of transactions classified as fraudulent.

Business implication: The current flagging approach represented in the dataset may not be sufficient. The company should consider improving its automated fraud-detection and risk-scoring capabilities.

Insight 5: Fraud prevention should be risk-based

The analysis suggests that fraudulent transactions are disproportionately associated with high-value transactions.

Business implication: Instead of applying the same controls to every transaction, the company should prioritize transactions based on amount, transaction type, account behavior, and other risk indicators.
